# Amazon Bestseller Rank Tracker with Scavio

Track the current bestsellers in any Amazon category and understand why they win, using the Scavio Amazon API and LangChain. A free alternative to Jungle Scout / Helium 10 for quick category scans.

**What you will learn:**
- Search Amazon sorted by bestsellers with `ScavioAmazonSearch`
- Read price, rating, and review-volume signals
- Have an LLM explain the category leaders

```python
# pip install langchain langchain-openai "langchain-scavio>=2.9" python-dotenv
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioAmazonSearch

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are an Amazon category analyst. For the given category:
1. Call ScavioAmazonSearch (favor bestseller/featured ordering).
2. List the top 8 products: title, price, rating, reviews_count, and the
   'best_seller'/'amazons_choice' badges when present.
3. Explain the 3 patterns that the leaders share (price band, rating bar,
   review volume). Use only returned data; never invent numbers."""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4o-mini", temperature=0)
    tools = [ScavioAmazonSearch(max_results=12)]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({"messages": [{"role": "user", "content": 'stainless steel water bottle'}]})
print(result["messages"][-1].content)
```
