# YouTube Video Research Agent with Scavio API

Search YouTube videos and analyze engagement metrics, channel data, and content trends using the Scavio search API and LangChain. Find trending creators and breakout videos in any niche.

**What you will learn:**
- Search YouTube videos with ScavioYouTubeSearch
- Pull detailed video metadata (views, likes, description) with ScavioYouTubeMetadata
- Analyze video engagement and identify trending content
- Build a video research report

**Prerequisites:**
- Free Scavio API key (250 credits/month): https://dashboard.scavio.dev
- OpenAI API key

**Tools used:** ScavioYouTubeSearch, ScavioYouTubeMetadata

```python
# pip install langchain langchain-openai langchain-scavio python-dotenv
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioYouTubeSearch, ScavioYouTubeMetadata

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are YouTubeResearchAgent, a video research and analysis agent.

Workflow:
1. Take the user's topic or niche.
2. Call ScavioYouTubeSearch to find the top videos on this topic.
3. Call ScavioYouTubeMetadata on the top 2-3 video IDs to get full
   details: title, description, view count, like count, channel, tags.
4. Analyze the results and produce:

   ## YouTube Research Report: <topic>

   ### Top Videos
   For each video:
   - **<title>**
     - Channel: <channel name>
     - Views: <view count> | Likes: <like count>
     - Published: <date>
     - Key topics from description: <2-3 bullet points>

   ### Content Trends
   - Common themes across top videos
   - Gaps: topics mentioned in descriptions but not deeply covered
   - Tags that appear across multiple videos

   ### Creator Insights
   - Which channels dominate this topic
   - Engagement patterns (like-to-view ratios)

   ### Content Opportunity
   2-3 video ideas based on gaps and trends found.

Rules:
- Never invent view counts, titles, or channel names. Only use tool output.
- Call only ONE tool per step.
- Keep the final report under 400 words.
"""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4.1-mini", temperature=0)
    tools = [
        ScavioYouTubeSearch(max_results=5),
        ScavioYouTubeMetadata(),
    ]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({
    "messages": [{"role": "user", "content": "how to start a SaaS business in 2026"}]
})
print(result["messages"][-1].content)
```
