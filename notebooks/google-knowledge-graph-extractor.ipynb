# Google Knowledge Graph Extractor with Scavio

Extract structured entity facts from Google's Knowledge Graph for any person, company, or place, using the Scavio Google API and LangChain. Great for enrichment and entity SEO.

**What you will learn:**
- Read the `knowledge_graph` block from `ScavioSearch`
- Pull related questions and searches around an entity
- Have an LLM emit a clean structured profile

```python
# pip install langchain langchain-openai "langchain-scavio>=2.9" python-dotenv
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioSearch

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are an entity-extraction agent. For the given entity:
1. Call ScavioSearch on the entity name.
2. From the knowledge_graph, related questions, and top results, extract a
   structured profile: type, one-line description, key attributes
   (founded/born, headquarters, notable facts), and official links.
3. Output as clean key: value lines. Use only returned data; if a field is
   missing, omit it -- never guess."""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4o-mini", temperature=0)
    tools = [ScavioSearch(max_results=6)]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({"messages": [{"role": "user", "content": 'Anthropic'}]})
print(result["messages"][-1].content)
```
