# TikTok Hashtag Analyzer with Scavio API

Look up TikTok hashtags, pull top videos, and analyze engagement and comment sentiment using the Scavio search API and LangChain. A free alternative to Pentos, Exolyt, and TikTok Analytics Pro.

**What you will learn:**
- Look up TikTok hashtag data with ScavioTikTokHashtag
- Pull top videos under a hashtag with ScavioTikTokHashtagVideos
- Analyze video comments with ScavioTikTokVideoComments
- Build a hashtag performance report

**Prerequisites:**
- Free Scavio API key (50 free credits (one-time)): https://dashboard.scavio.dev
- OpenAI API key

**Tools used:** ScavioTikTokHashtag, ScavioTikTokHashtagVideos, ScavioTikTokVideoComments

```python
# pip install langchain langchain-openai langchain-scavio python-dotenv
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import (
    ScavioTikTokHashtag,
    ScavioTikTokHashtagVideos,
    ScavioTikTokVideoComments,
)

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are TikTokHashtagAnalyzer, a hashtag research agent.

Workflow:
1. Take the user's hashtag or topic.
2. Call ScavioTikTokHashtag with the hashtag name to get hashtag info
   and the hashtag ID.
3. Call ScavioTikTokHashtagVideos with the hashtag ID to get top videos.
4. Call ScavioTikTokVideoComments on the top video to analyze
   comment sentiment.
5. Compile a hashtag analysis report:

   ## TikTok Hashtag Report: #<hashtag>

   ### Hashtag Overview
   - Hashtag ID: <id>
   - Total views/posts if available

   ### Top Videos
   For each of the top 3-5 videos:
   - **<video description snippet>**
     - Creator: @<username>
     - Likes: <count> | Comments: <count> | Shares: <count>

   ### Comment Analysis
   From the top video's comments:
   - Overall sentiment: <positive/negative/mixed>
   - Common themes in comments
   - Notable comments or questions

   ### Hashtag Insights
   - Best posting format (duration, style, hook type)
   - Engagement patterns
   - Audience interests based on comments

   ### Recommendation
   How to use this hashtag effectively for marketing or content creation.

Rules:
- Never invent usernames, view counts, or comment text. Only use tool output.
- Call only ONE tool per step.
- Keep the report under 400 words.
"""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4.1-mini", temperature=0)
    tools = [
        ScavioTikTokHashtag(),
        ScavioTikTokHashtagVideos(max_results=5),
        ScavioTikTokVideoComments(max_results=10),
    ]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({
    "messages": [{"role": "user", "content": "sidehustle"}]
})
print(result["messages"][-1].content)
```
