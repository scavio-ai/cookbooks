# SEO Keyword Researcher with Scavio API

Extract People Also Ask questions, related searches, and knowledge graph data for any keyword using the Scavio search API and LangChain. A free alternative to Ahrefs, Semrush, and Moz keyword tools.

**What you will learn:**
- Search Google with ScavioSearch to get structured SEO data
- Extract People Also Ask (PAA) questions for content ideas
- Mine related searches for keyword clustering
- Build a keyword research report with an LLM

**Prerequisites:**
- Free Scavio API key (250 credits/month): https://dashboard.scavio.dev
- OpenAI API key

**Tools used:** ScavioSearch

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioSearch

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are SEOKeywordResearcher, a keyword research agent.

Workflow:
1. Take the user's seed keyword.
2. Call ScavioSearch with the seed keyword. Extract:
   - Top 5 organic results (title, URL, snippet)
   - All People Also Ask questions
   - All related searches
3. Call ScavioSearch on the 2 most promising related searches to expand
   the keyword cluster.
4. Compile a keyword research report:

   ## Seed Keyword: <keyword>

   ### Top-Ranking Pages
   For each of the top 5 results: title, URL, one-line summary.

   ### People Also Ask (Content Ideas)
   Numbered list of every PAA question found. These are blog post titles.

   ### Related Keywords
   Bulleted list of all related searches across queries.

   ### Recommended Content Plan
   3-5 article ideas based on PAA gaps (questions where current top
   results give weak answers). Each idea: title + target keyword + angle.

Rules:
- Never invent URLs, titles, or questions. Only use tool output.
- Call only ONE tool per step.
- Keep the final report under 400 words.
"""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4.1-mini", temperature=0)
    tools = [ScavioSearch(max_results=5)]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({
    "messages": [{"role": "user", "content": "best creatine supplements 2026"}]
})
print(result["messages"][-1].content)
```
