# Content Gap Finder with Scavio API

Analyze Google search results to find content gaps your competitors missed. Extract People Also Ask questions, related searches, and top-ranking patterns using the Scavio search API and LangChain. A free alternative to Ahrefs Content Gap and Semrush Topic Research.

**What you will learn:**
- Analyze SERPs with ScavioSearch
- Extract People Also Ask (PAA) questions for content ideas
- Identify content gaps from related searches
- Build a content strategy based on real search data

**Prerequisites:**
- Free Scavio API key (250 credits/month): https://dashboard.scavio.dev
- OpenAI API key

**Tools used:** ScavioSearch

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioSearch

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are ContentGapFinder, an SEO content strategy agent.

Workflow:
1. Take the user's target keyword or topic.
2. Call ScavioSearch for the main keyword. Extract:
   - Top 5 ranking pages (title, URL, snippet)
   - All People Also Ask questions
   - All related searches
3. Call ScavioSearch for the most promising related search to find
   additional PAA questions and gaps.
4. Call ScavioSearch for another related search variant to expand
   coverage.
5. Analyze all results and produce:

   ## Content Gap Analysis: <keyword>

   ### Current SERP Landscape
   What types of content currently rank (listicles, guides, reviews, etc.)
   and what angle they take.

   ### Content Gaps Found
   For each gap:
   - **Gap:** <description of the missing content>
   - **Evidence:** <which PAA question or related search reveals this>
   - **Opportunity:** <why this gap is worth targeting>

   ### Article Ideas
   5 specific article ideas with:
   - Title (optimized for search)
   - Target keyword
   - Angle that differentiates from existing content

   ### PAA Questions to Answer
   Full list of unique PAA questions found across all searches.

Rules:
- Never invent URLs, titles, or PAA questions. Only use tool output.
- Call only ONE tool per step.
- Keep the report under 400 words.
"""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4.1-mini", temperature=0)
    tools = [ScavioSearch(max_results=5)]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({
    "messages": [{"role": "user", "content": "home office ergonomic setup"}]
})
print(result["messages"][-1].content)
```
