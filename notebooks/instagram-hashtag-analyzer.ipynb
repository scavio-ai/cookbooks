# Instagram Hashtag Analyzer with Scavio

Find and rank Instagram hashtags by volume for any niche, and spot the content opportunity, using the Scavio Instagram search API and LangChain. A free alternative to paid Instagram hashtag tools.

**What you will learn:**
- Search Instagram hashtags with `ScavioInstagramSearchHashtags`
- Rank hashtags by media volume to size a niche
- Have an LLM cluster hashtags into a posting strategy

**Prerequisites:** free Scavio API key (https://dashboard.scavio.dev) + OpenAI key.

```python
# pip install langchain langchain-openai "langchain-scavio>=2.9" python-dotenv
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioInstagramSearchHashtags

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are an Instagram hashtag strategist. Given a niche:
1. Call ScavioInstagramSearchHashtags on 2-3 seed terms.
2. From the returned hashtags and their media counts, group them into
   BIG (broad reach), MEDIUM (niche), and SMALL (easy to rank) buckets.
3. Recommend a 5-7 hashtag mix for a new account, with one line of why.
Use only hashtags returned by the tool; never invent counts."""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4o-mini", temperature=0)
    tools = [ScavioInstagramSearchHashtags(max_results=15)]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({"messages": [{"role": "user", "content": 'specialty coffee'}]})
print(result["messages"][-1].content)
```
