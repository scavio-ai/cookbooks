# Walmart Deal Finder with Scavio API

Search Walmart for deals in any category, pull full product details, and rank by value using the Scavio search API and LangChain. The only open-source Walmart API cookbook on GitHub.

**What you will learn:**
- Search Walmart products by keyword with ScavioWalmartSearch
- Pull detailed product data with ScavioWalmartProduct
- Rank deals by price-to-rating ratio
- Find the best value products on Walmart

**Prerequisites:**
- Free Scavio API key (250 credits/month): https://dashboard.scavio.dev
- OpenAI API key

**Tools used:** ScavioWalmartSearch, ScavioWalmartProduct

```python
# pip install langchain langchain-openai langchain-scavio python-dotenv
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioWalmartSearch, ScavioWalmartProduct

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are WalmartDealFinder, a Walmart shopping research agent.

Workflow:
1. Take the user's product category or search query.
2. Call ScavioWalmartSearch to find the top 5 products.
3. Call ScavioWalmartProduct on the top 3 results for full details
   (price, ratings, seller, availability).
4. Rank by value and compile a deal report:

   ## Walmart Deals: <category>

   ### Top Deals
   For each product:
   - **<Product Name>**
     - Price: $X.XX
     - Rating: X.X stars (X reviews)
     - Seller: <seller name>
     - Why it is a deal: <one-line reasoning>

   ### Best Overall Deal
   <product name> -- explain why based on price, rating, and reviews.

   ### Budget Pick
   <cheapest product with acceptable rating>

   ### Skip
   Any product with red flags (low rating, few reviews, suspicious seller).

Rules:
- Never invent prices, ratings, or product names. Only use tool output.
- Call only ONE tool per step.
- Keep the final report under 300 words.
"""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4.1-mini", temperature=0)
    tools = [
        ScavioWalmartSearch(max_results=5),
        ScavioWalmartProduct(),
    ]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({
    "messages": [{"role": "user", "content": "air fryer under $80"}]
})
print(result["messages"][-1].content)
```
