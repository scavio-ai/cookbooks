# YouTube to Blog Post Converter with Scavio API

Convert any YouTube video into an SEO-optimized blog post outline using video metadata and Google search data. Uses the Scavio search API and LangChain to combine YouTube content signals with real SERP data for SEO.

**What you will learn:**
- Pull video metadata (title, description, tags) with ScavioYouTubeMetadata
- Extract People Also Ask questions with ScavioSearch for SEO headings
- Generate a blog post outline optimized for search

**Prerequisites:**
- Free Scavio API key (250 credits/month): https://dashboard.scavio.dev
- OpenAI API key

**Tools used:** ScavioYouTubeSearch, ScavioYouTubeMetadata, ScavioSearch

```python
# pip install langchain langchain-openai langchain-scavio python-dotenv
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioYouTubeSearch, ScavioYouTubeMetadata, ScavioSearch

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are YouTubeToBlogPost, a content repurposing agent.

Workflow:
1. Take the user's YouTube topic or search query.
2. Call ScavioYouTubeSearch to find the most relevant video.
3. Call ScavioYouTubeMetadata on the top video ID to get the full
   title, description, tags, view count, and channel info.
4. Call ScavioSearch for the video's main topic to get:
   - People Also Ask questions (use as H2/H3 headings)
   - Related searches (use as subtopics)
   - Top-ranking pages (understand what already ranks)
5. Generate a blog post outline:

   ## Blog Post Outline: <SEO-optimized title>

   ### Source Video
   - Title: <video title>
   - Channel: <channel>
   - Views: <view count>

   ### SEO Title
   <optimized title using keywords from Google search data>

   ### Meta Description
   <150-character description targeting the main keyword>

   ### Outline
   For each section:
   - H2: <heading based on PAA question or video topic>
     - Key points to cover (from video description)
     - Supporting data or examples

   ### People Also Ask (Use as FAQ Section)
   List all PAA questions found -- each becomes a FAQ entry.

   ### Target Keywords
   List of keywords from related searches and tags.

Rules:
- Never invent video data, PAA questions, or URLs. Only use tool output.
- Call only ONE tool per step.
- Keep the outline under 400 words.
"""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4.1-mini", temperature=0)
    tools = [
        ScavioYouTubeSearch(max_results=3),
        ScavioYouTubeMetadata(),
        ScavioSearch(max_results=5),
    ]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({
    "messages": [{"role": "user", "content": "morning routine for productivity"}]
})
print(result["messages"][-1].content)
```
