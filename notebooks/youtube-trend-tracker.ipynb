# YouTube Trend Tracker with Scavio API

Track trending YouTube videos in any niche, analyze view velocity, and identify rising creators using the Scavio search API and LangChain. A free alternative to VidIQ and TubeBuddy trend tools.

**What you will learn:**
- Search YouTube by topic with ScavioYouTubeSearch
- Analyze video metadata (views, likes, publish date) with ScavioYouTubeMetadata
- Calculate view velocity to identify breakout content
- Build a trend report for any niche

**Prerequisites:**
- Free Scavio API key (250 credits/month): https://dashboard.scavio.dev
- OpenAI API key

**Tools used:** ScavioYouTubeSearch, ScavioYouTubeMetadata

```python
# pip install langchain langchain-openai langchain-scavio python-dotenv
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_scavio import ScavioYouTubeSearch, ScavioYouTubeMetadata

load_dotenv(override=True)
```

```python
SYSTEM_PROMPT = """You are YouTubeTrendTracker, a video trend analysis agent.

Workflow:
1. Take the user's topic or niche.
2. Call ScavioYouTubeSearch for the topic to find recent videos.
3. Call ScavioYouTubeMetadata on the top 3 video IDs for full details:
   title, channel, view count, like count, publish date, description.
4. Analyze trends and produce:

   ## YouTube Trend Report: <topic>

   ### Trending Videos
   For each video (ranked by engagement):
   - **<title>**
     - Channel: <channel>
     - Views: <count> | Likes: <count>
     - Published: <date>
     - Like/View ratio: <percentage>

   ### Trend Analysis
   - Dominant content format (tutorial, review, vlog, explainer)
   - Common themes in titles and descriptions
   - Which channels are growing fastest in this niche

   ### Rising Signals
   - Videos with high like/view ratios (>5%) indicate strong engagement
   - Recent uploads with high view counts indicate trending topics

   ### Content Opportunity
   2-3 video ideas based on what is trending but underserved.

Rules:
- Never invent view counts, titles, or channel names. Only use tool output.
- Call only ONE tool per step.
- Keep the report under 350 words.
"""
```

```python
def build_agent():
    model = ChatOpenAI(model="gpt-4.1-mini", temperature=0)
    tools = [
        ScavioYouTubeSearch(max_results=5),
        ScavioYouTubeMetadata(),
    ]
    return create_agent(model, tools=tools, system_prompt=SYSTEM_PROMPT)
```

```python
agent = build_agent()
result = agent.invoke({
    "messages": [{"role": "user", "content": "AI coding tools and agents"}]
})
print(result["messages"][-1].content)
```
